```python
LABEL_MAP = {
    "GIVENNAME":        "PERSON",
    "SURNAME":          "PERSON",
    "USERNAME":         "PERSON",
    "EMAIL":            "EMAIL",
    "TELEPHONENUM":     "PHONE",
    "CREDITCARDNUMBER": "CREDIT_CARD",
    "ACCOUNTNUM":       "CREDIT_CARD",
    "PASSWORD":         "DB_CREDENTIAL",
    "TAXNUM":           "PAN",
    "IDCARDNUM":        "AADHAAR",
    "SOCIALNUM":        "AADHAAR",
    "DRIVERLICENSENUM": "AADHAAR",
}

SKIP_LABELS = {
    "BUILDINGNUM", "CITY", "STREET",
    "ZIPCODE", "DATEOFBIRTH"
}

def extract_example(row):
    """
    Convert one ai4privacy row into our format.
    Returns None if no spans match our categories.
    """
    source = row["source_text"]
    spans = []

    for mask in row["privacy_mask"]:
        label = mask["label"]
        if label in SKIP_LABELS:
            continue
        if label not in LABEL_MAP:
            continue
        spans.append({
            "text":     mask["value"],
            "category": LABEL_MAP[label]
        })

    # Skip if no relevant spans found
    if len(spans) == 0:
        return None

    # Skip if source text looks synthetic/broken
    # (some entries have placeholder text like *BF6* everywhere)
    if source.count("*") > 6:
        return None

    return {
        "source":   source,
        "spans":    spans,
        "origin":   "ai4privacy"
    }

# Load a larger sample to get enough after filtering
print("Loading dataset...")
raw = load_dataset("ai4privacy/pii-masking-400k", split="train[:5000]")
print(f"Loaded {len(raw)} raw examples")

# Extract and filter
examples = []
for row in raw:
    result = extract_example(row)
    if result:
        examples.append(result)

print(f"Kept {len(examples)} examples after filtering")
print("\nSample example:")
print(json.dumps(examples[0], indent=2))
```

```python
# CORRECTED label map — only mappings where values actually match
LABEL_MAP = {
    "GIVENNAME":         "PERSON",
    "SURNAME":           "PERSON",
    "USERNAME":          "PERSON",
    "EMAIL":             "EMAIL",
    "TELEPHONENUM":      "PHONE",
    "CREDITCARDNUMBER":  "CREDIT_CARD",
    # Removed: IDCARDNUM, SOCIALNUM, DRIVERLICENSENUM, PASSWORD, TAXNUM, ACCOUNTNUM
}

SKIP_LABELS = {
    "BUILDINGNUM", "CITY", "STREET", "ZIPCODE",
    "DATEOFBIRTH", "IDCARDNUM", "SOCIALNUM",
    "DRIVERLICENSENUM", "PASSWORD", "TAXNUM", "ACCOUNTNUM"
}

def extract_example(row):
    source = row["source_text"]
    spans = []

    for mask in row["privacy_mask"]:
        label = mask["label"]
        if label in SKIP_LABELS:
            continue
        if label not in LABEL_MAP:
            continue
        spans.append({
            "text":     mask["value"],
            "category": LABEL_MAP[label]
        })

    if len(spans) == 0:
        return None

    # Filter out heavily synthetic entries
    if source.count("*") > 6:
        return None

    # Filter out HTML-heavy entries
    if source.count("<") > 8:
        return None

    return {
        "source":  source,
        "spans":   spans,
        "origin":  "ai4privacy"
    }

# Re-extract with corrected map
print("Re-extracting with corrected label map...")
raw = load_dataset("ai4privacy/pii-masking-400k", split="train[:5000]")

examples_clean = []
for row in raw:
    result = extract_example(row)
    if result:
        examples_clean.append(result)

print(f"Kept {len(examples_clean)} examples after corrected filtering")

# Verify — spot check PERSON, EMAIL, PHONE, CREDIT_CARD only
from collections import Counter
all_cats = []
for ex in examples_clean:
    for span in ex["spans"]:
        all_cats.append(span["category"])

print("\nCategory distribution after fix:")
for cat, count in Counter(all_cats).most_common():
    print(f"  {cat}: {count}")

# Spot check one of each
print("\n=== Sample PERSON ===")
for ex in examples_clean:
    for span in ex["spans"]:
        if span["category"] == "PERSON":
            print(f"  Value: {span['text']}")
            print(f"  Source snippet: {ex['source'][:100]}")
            break
    else:
        continue
    break

print("\n=== Sample EMAIL ===")
for ex in examples_clean:
    for span in ex["spans"]:
        if span["category"] == "EMAIL":
            print(f"  Value: {span['text']}")
            print(f"  Source snippet: {ex['source'][:100]}")
            break
    else:
        continue
    break

print("\n=== Sample CREDIT_CARD ===")
for ex in examples_clean:
    for span in ex["spans"]:
        if span["category"] == "CREDIT_CARD":
            print(f"  Value: {span['text']}")
            print(f"  Source snippet: {ex['source'][:100]}")
            break
    else:
        continue
    break
```

```python
# Drop PERSON from public dataset, keep only clean universal categories
examples_public = []

for ex in examples_clean:
    # Remove PERSON spans from public dataset examples
    clean_spans = [
        span for span in ex["spans"]
        if span["category"] != "PERSON"
    ]

    # Only keep example if it still has at least one span
    if len(clean_spans) == 0:
        continue

    examples_public.append({
        "source": ex["source"],
        "spans":  clean_spans,
        "origin": "ai4privacy"
    })

print(f"Public dataset examples after dropping PERSON: {len(examples_public)}")

# Verify distribution
from collections import Counter
all_cats = []
for ex in examples_public:
    for span in ex["spans"]:
        all_cats.append(span["category"])

print("\nFinal public dataset category distribution:")
for cat, count in Counter(all_cats).most_common():
    print(f"  {cat}: {count}")

# Take a random sample of 600 from public dataset
import random
random.seed(42)
public_sample = random.sample(
    examples_public,
    min(600, len(examples_public))
)
print(f"\nPublic sample taken: {len(public_sample)}")
```
