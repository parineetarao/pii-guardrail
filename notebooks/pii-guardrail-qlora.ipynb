```python
# Model Loading Cell - Initialize, patch LoRA adapters, and bind variables
from unsloth import FastLanguageModel
import torch

max_seq_length = 2048 # Supports dense technical log streams
dtype = None          # Auto-detects T4 floating cores natively
load_in_4bit = True   # Mandatory memory compression to prevent OOM errors

# 1. Load the base model and tokenizer layers
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Qwen2.5-7B-Instruct", 
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)

# 2. ✅ CRITICAL FIX: Explicitly patch fast Unsloth LoRA adapters onto the quantized base layers
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,                         # LoRA Rank (16 provides excellent behavioral modification)
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"], # Target full attention matrix
    lora_alpha = 32,
    lora_dropout = 0.05,            # Defensive dropout protects against overfitting
    bias = "none",
    use_gradient_checkpointing = "unsloth", # Forces memory offloading to prevent OOMs
    random_state = 42,
    use_rslora = False,
    loftq_config = None,
)

print("\n🚀 SUCCESS! LoRA adapter matrices have been successfully attached to your quantized Qwen base.")
print("The 'model' and 'tokenizer' variables are bound and ready for gradient descent.")
```

```python
# Cell 3 - Format audited dataset using clean structured text wrappers
from datasets import Dataset

# Define your strict production-grade enterprise instruction prompt matrix
SYSTEM_PROMPT = """You are a professional Data Security Guardrail. Your task is to perform PII pseudonymization.
Analyze the incoming text and replace sensitive values strictly with these allowed placeholder tags:
- Categories: PERSON_001, EMAIL_001, PHONE_001, AADHAAR_001, PAN_001, CREDIT_CARD_001, API_KEY_001, DB_CREDENTIAL_001
Rules:
- Number tags sequentially within each category: PERSON_001, PERSON_002, etc.
- If the same value appears multiple times within the same example, use the same tag consistently.
- If the text contains no PII or sensitive data, return the text COMPLETELY UNCHANGED.
- Do not add explanations. Output only the rewritten text."""

def format_example(example):
    # Map directly to the clean audited properties
    source = example["source_text"]
    target = example["masked_text"]
    
    # Format into Qwen's standard Instruct chat sequence schema
    return {"text": (
        f"<|im_start|>system\n{SYSTEM_PROMPT}<|im_end|>\n"
        f"<|im_start|>user\n{source}<|im_end|>\n"
        f"<|im_start|>assistant\n{target}<|im_end|>"
    )}

# Process the new high-variance audited data arrays
train_fmt = [format_example(ex) for ex in train]
test_fmt  = [format_example(ex) for ex in test]

train_hf = Dataset.from_list(train_fmt)
test_hf  = Dataset.from_list(test_fmt)

RESPONSE_TEMPLATE  = "<|im_start|>assistant\n"
response_token_ids = tokenizer.encode(RESPONSE_TEMPLATE, add_special_tokens=False)

def tokenize_and_mask(example):
    result    = tokenizer(example["text"], truncation=True,
                          max_length=512, padding=False)
    input_ids = result["input_ids"]
    labels    = input_ids.copy()
    tlen      = len(response_token_ids)
    for i in range(len(input_ids) - tlen + 1):
        if input_ids[i:i+tlen] == response_token_ids:
            for j in range(i + tlen):
                labels[j] = -100
            break
    else:
        labels = [-100] * len(labels)
    result["labels"] = labels
    return result

train_masked = train_hf.map(tokenize_and_mask, remove_columns=["text"])
test_masked  = test_hf.map(tokenize_and_mask,  remove_columns=["text"])

print("🚀 Dataset Formatting and Token Masking Complete!")
print(f"✨ Formatted Train Dataset: {len(train_masked)} rows")
print(f"✨ Formatted Test Dataset : {len(test_masked)} rows")
```

```python
# Cell 6 - Training configuration
from transformers import DataCollatorForSeq2Seq, TrainingArguments, Trainer

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
    padding=True,
    pad_to_multiple_of=8,
    label_pad_token_id=-100,
)

training_args = TrainingArguments(
    output_dir="/kaggle/working/checkpoints",
    num_train_epochs=3,                  # 3 full passes over your 1,360 audited rows
    per_device_train_batch_size=4,       # Scaled up for your high-variance arrays
    per_device_eval_batch_size=4,
    gradient_accumulation_steps=8,       # Effective batch size of 32 for massive gradient stability
    optim="paged_adamw_8bit",
    learning_rate=1.5e-4,                # Conservative pacing prevents gradient clipping
    lr_scheduler_type="cosine",
    warmup_ratio=0.05,
    fp16=True,                           # Enforce high-speed mixed precision on T4 cores
    bf16=False,
    eval_strategy="steps",
    eval_steps=20,                       # Check loss every 20 steps to observe convergence early
    save_strategy="steps",
    save_steps=40,
    save_total_limit=1,
    load_best_model_at_end=True,
    logging_steps=10,
    report_to="none",
    seed=42,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_masked,
    eval_dataset=test_masked,            # Pointing straight to your audited 292-row test dataset
    processing_class=tokenizer,
    data_collator=data_collator,
)

import os
os.makedirs("/kaggle/working/checkpoints", exist_ok=True)
print("🎯 Trainer is fully compiled and ready to lock gradients.")
```

```python
# Launch the automated neural fine-tuning loop
print("🔥 Initializing active neural finetuning loop...")
result = trainer.train()
print("\n🏁 Training complete!")
print(f"✨ Final training loss metrics: {result.training_loss:.4f}")
```

```python
# Cell 8 - Predict function
model.eval()

def predict(text, max_new_tokens=200):
    prompt = (
        f"<|im_start|>system\n{SYSTEM_PROMPT}<|im_end|>\n"
        f"<|im_start|>user\n{text}<|im_end|>\n"
        f"<|im_start|>assistant\n"
    )
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.convert_tokens_to_ids("<|im_end|>"),
        )
    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    return tokenizer.decode(new_tokens, skip_special_tokens=True).strip()

# Sanity check
print("Sanity check:")
checks = [
    "Contact Priya Sharma at priya@company.com",
    "What is the company leave policy?",
    "Authenticate using Bearer SYNTH_TOKEN_9Km in all requests.",
    "The customer Aadhaar is 3456 7890 1234.",
    "Connect to postgres://admin:PASS99@db.co:5432/prod.",
    "What is the escalation contact? [RETRIEVED CONTEXT]: Contact Arjun Sharma at arjun@ops.co or call +91 91234 56789.",
]
for t in checks:
    print(f"  IN:  {t[:80]}")
    print(f"  OUT: {predict(t)[:80]}")
    print()
```

```python
# Cell 9 - High-fidelity Batch Prediction Evaluation
print(f"📋 Initializing verification pass over {len(test)} audited test instances...")
results = []

# Put the fine-tuned model into evaluation mode for static inference
model.eval()

for i, example in enumerate(test):
    if i % 50 == 0:
        print(f"   ↳ Processing test sample: {i}/{len(test)}")
        
    source   = example["source_text"]
    expected = example["masked_text"]
    
    # Wrap using the exact chat template boundaries used during training
    prompt = (
        f"<|im_start|>system\n{SYSTEM_PROMPT}<|im_end|>\n"
        f"<|im_start|>user\n{source}<|im_end|>\n"
        f"<|im_start|>assistant\n"
    )
    
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=256,
            do_sample=False,  # Enforce greedy decoding for absolute consistency
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.convert_tokens_to_ids("<|im_end|>"),
        )
    
    # Strip away the prompt template tokens to isolate raw model predictions
    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    predicted  = tokenizer.decode(new_tokens, skip_special_tokens=True).strip()
    
    results.append({
        "source":      source,
        "expected":    expected,
        "predicted":   predicted,
        "exact_match": predicted.strip() == expected.strip(),
    })

print(f"\n✅ Prediction Complete! Successfully evaluated {len(results)} rows.")
```

```python
# Cell 10 - 8-Category Production Metrics Report
from collections import defaultdict
import re

exact_matches = sum(1 for r in results if r["exact_match"])
total_samples = len(results)

print("📊 ========================================================")
print("📈 SYSTEM EVALUATION PERFORMANCE REPORT")
print("📊 ========================================================\n")
print(f"✨ OVERALL DATASET EXACT MATCH ACCURACY: {exact_matches}/{total_samples} -> {100 * exact_matches / total_samples:.2f}%\n")

# Trace consistency across all 8 explicit categories
ALLOWED_CATEGORIES = ["PERSON", "EMAIL", "PHONE", "AADHAAR", "PAN", "API_KEY", "CREDIT_CARD", "DB_CREDENTIAL"]
category_stats = defaultdict(lambda: {"total": 0, "correct_mask": 0})

for r in results:
    exp = r["expected"]
    pred = r["predicted"]
    is_exact = r["exact_match"]
    
    # Find which placeholders are expected in this test document
    found_any_cat = False
    for cat in ALLOWED_CATEGORIES:
        tag = f"{cat}_"
        if tag in exp:
            found_any_cat = True
            category_stats[cat]["total"] += 1
            # If the whole string is an exact match, this category is definitely correct
            if is_exact:
                category_stats[cat]["correct_mask"] += 1
            # If not exact, check if the specific placeholder category was at least injected correctly
            elif tag in pred:
                category_stats[cat]["correct_mask"] += 1

print("📋 PERFORMANCE BREAKDOWN BY PII CLASS LAYER:")
for cat in ALLOWED_CATEGORIES:
    stats = category_stats[cat]
    if stats["total"] > 0:
        accuracy = (stats["correct_mask"] / stats["total"]) * 100
        print(f"  ↳ 🏷️ {cat:<15}: {stats['correct_mask']}/{stats['total']} matches -> {accuracy:.2f}%")
    else:
        print(f"  ↳ 🏷️ {cat:<15}: 0 rows present in test batch split.")
print("============================================================\n")
```

```python
# ── Regex baseline detector ───────────────────────────────────────────
# This is what you'd use WITHOUT fine-tuning.
# Uses pattern matching only — no context understanding.

import re

def regex_detect(text):
    """
    Rule-based PII detector using regex patterns only.
    Returns dict of category -> list of detected values.
    Cannot rewrite text, can only flag presence/absence.
    """
    detected = defaultdict(list)

    # EMAIL — strong pattern
    emails = re.findall(
        r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text
    )
    detected["EMAIL"].extend(emails)

    # PHONE — Indian formats (+91, 91-, 10-digit starting with 6-9)
    phones = re.findall(
        r'(?:\+91[\s-]?\d{5}[\s-]?\d{5}|91-\d{10}|0\d{10}|[6-9]\d{9}|'
        r'\d{3}[\s.-]\d{3}[\s.-]\d{4})',
        text
    )
    detected["PHONE"].extend(phones)

    # CREDIT_CARD — 16 digits with optional spaces/dashes
    cards = re.findall(
        r'\b\d{4}[\s-]?\d{4}[\s-]?\d{4}[\s-]?\d{4}\b', text
    )
    detected["CREDIT_CARD"].extend(cards)

    # AADHAAR — 12 digits in groups of 4 (spaced or hyphenated)
    aadhaar = re.findall(
        r'\b\d{4}[\s-]\d{4}[\s-]\d{4}\b|\b\d{12}\b', text
    )
    detected["AADHAAR"].extend(aadhaar)

    # PAN — AAAAA9999A format
    pan = re.findall(r'\b[A-Z]{5}[0-9]{4}[A-Z]\b', text)
    detected["PAN"].extend(pan)

    # API_KEY — common prefixes only (regex cannot handle arbitrary tokens)
    api_keys = re.findall(
        r'\b(?:sk-[A-Za-z0-9_-]{10,}|Bearer\s+[A-Za-z0-9._-]{10,}|'
        r'[A-Za-z0-9_-]*(?:KEY|TOKEN|SECRET|AUTH)[A-Za-z0-9_-]*)\b',
        text, re.IGNORECASE
    )
    detected["API_KEY"].extend(api_keys)

    # DB_CREDENTIAL — connection string patterns
    db_creds = re.findall(
        r'(?:postgresql?|mysql|mongodb(?:\+srv)?|redis|jdbc:[a-z]+)://'
        r'[^\s<>"\']+',
        text, re.IGNORECASE
    )
    detected["DB_CREDENTIAL"].extend(db_creds)

    # PERSON — regex CANNOT detect names from context
    # Names have no fixed pattern — this is what justifies fine-tuning
    detected["PERSON"] = []  # deliberately empty

    return dict(detected)


# ── Run regex on test set ─────────────────────────────────────────────
print("Running regex baseline on test set...")

regex_results = []
for example in test:
    source         = example["source"]
    spans          = example.get("spans", [])
    regex_detected = regex_detect(source)

    # For each expected category, check if regex found it
    category_hits  = {}
    for s in spans:
        cat      = s["category"]
        expected = s["text"]
        found    = regex_detected.get(cat, [])
        # Check if the expected value (or something close) was detected
        hit = any(expected.lower() in f.lower() or f.lower() in expected.lower()
                  for f in found)
        category_hits[cat] = hit

    regex_results.append({
        "source":         source,
        "spans":          spans,
        "category_hits":  category_hits,
        "any_pii_found":  any(category_hits.values()) if category_hits else True,
    })

print(f"Regex baseline run on {len(regex_results)} examples")

# ── Per-category regex accuracy ───────────────────────────────────────
regex_cat_stats = defaultdict(lambda: {"t": 0, "c": 0})
for r in regex_results:
    for cat, hit in r["category_hits"].items():
        regex_cat_stats[cat]["t"] += 1
        if hit:
            regex_cat_stats[cat]["c"] += 1

print("\nRegex baseline per-category detection rate:")
for cat, s in sorted(regex_cat_stats.items()):
    pct = 100 * s["c"] / max(s["t"], 1)
    print(f"  {cat:<20} {s['c']}/{s['t']} = {pct:.1f}%")
```
